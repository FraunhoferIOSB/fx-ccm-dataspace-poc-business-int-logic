```python
# === READ Shells ===
url = BASE_URL + "/shells"  # use http or https as needed
headers = {"Content-Type": "application/json", "Accept": "application/json"}


# get existing shells
try:
    response = requests.get(url, headers=headers, verify=False)
    response.raise_for_status()
    data = response.json()
    print("Raw JSON response:")
    print(data)
    with open("shells_response.json", "w", encoding="utf-8") as f:
        json.dump(data, f, indent=2)
    
    print("\nParsed Shell IDs/Names:")
    for shell in data.get('result', []):
        print(f"ID: {shell.get('idShort', 'N/A')} | Identifier: {shell.get('identification', {}).get('id', 'N/A')}")
except requests.exceptions.RequestException as e:
    print(f"Request failed: {e}")
```
