```python
import requests
import urllib3
urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)
```

```python
url = "http://localhost:8000/api/v3.0/shells"  # use http or https as needed
headers = {"Content-Type": "application/json", "Accept": "application/json"}


# get existing shells
try:
    response = requests.get(url, headers=headers, verify=False)
    response.raise_for_status()
    data = response.json()
    print("Raw JSON response:")
    print(data)
    print("\nParsed Shell IDs/Names:")
    for shell in data.get('result', []):
        print(f"ID: {shell.get('idShort', 'N/A')} | Identifier: {shell.get('identification', {}).get('id', 'N/A')}")
except requests.exceptions.RequestException as e:
    print(f"Request failed: {e}")
```

```python
# === UPLOAD ===
import base64
import json

# Base URL of your AAS server (adjust host/port/prefix if needed)
BASE_URL = "https://localhost:8000/api/v3.0"

# Load your AAS JSON for Component A (as built from the template)
with open("env_axis.json", "r") as f:   # possibly add encoding: encoding="utf-8"
    aas_payload = json.load(f)


# Upload AAS (POST to /shells):
for aas_json in aas_payload['assetAdministrationShells']:
    print("Uploading AAS:", aas_json.get('idShort', 'N/A'))
    resp = requests.post(
        f"{BASE_URL}/shells",
        headers=headers,
        verify=False,
        data=json.dumps(aas_json)
    )
    print("Create AAS status:", resp.status_code, resp.text)



# Upload Submodels:
for submodel_json in aas_payload['submodels']:
    print("Uploading Submodel:", submodel_json.get('idShort', 'N/A'))
    resp = requests.post(
        f"{BASE_URL}/submodels",
        headers=headers,
        verify=False,
        data=json.dumps(submodel_json)
    )
    print("Create Submodel status:", resp.status_code, resp.text)
```
